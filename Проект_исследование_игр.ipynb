3.1 Основной фактор — эти данные помогут построить прогноз на 2017 год.

```python
plt.figure(figsize=(12, 15))

years = np.arange(2012, 2017)  # Данные за 2012–2016
future_year = 2017

# Получаем суммарные продажи по годам для платформы
for name in platform_sales:
    sales_by_year = (actual_games.query('platform == @name').pivot_table(index='year_of_release', values='total_sales', aggfunc='sum').reindex(years, fill_value=0)['total_sales'])
    
# Строим график фактических значений
    plt.plot(years, sales_by_year, marker='o', label=f'{name} (факт)')
    
# Линейная регрессия (прямая)
    coeffs = np.polyfit(years, sales_by_year, 1)
    trend = np.poly1d(coeffs)
    pred_2017 = trend(future_year)
    
# Добавляем прогнозную точку на 2017 год
    plt.scatter([future_year], [pred_2017], marker='*', s=150, label=f'{name} (прогноз 2017)')
    plt.plot([years[-1], future_year], [sales_by_year.iloc[-1], pred_2017], 'k--', alpha=0.3)

plt.title('Прогноз суммарных продаж по платформам на 2017 год')
plt.xlabel('Год')
plt.ylabel('Суммарные продажи')
plt.xticks(np.arange(2012, 2018, step=1))
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='upper right', fontsize=8)
plt.tight_layout()
plt.show()
```

Выдвинем гипотезы:

    a. Средние пользовательские рейтинги платформ XOne и PC равны.
    b. Средние пользовательские рейтинги платформ XOne и PC не равны.
    

```python

xone_game = actual_games.query('platform == "XOne" and user_score > 0')
print('Описательная статистика пользовательских рейтингов по XOne\n\n', xone_game['user_score'].describe())

pc_game = actual_games.query("platform == 'PC' and user_score > 0")
print('Описательная статистика пользовательских рейтингов по PC\n\n', pc_game['user_score'].describe())
```

Проведение t-теста Стьюдента для независимых выборок

```python
alpha = 0.05


ttest_res = st.ttest_ind(xone_game['user_score'],pc_game['user_score'],equal_var=False,alternative='two-sided')

p_value = ttest_res.pvalue
print(f'p-value: {p_value:.5f}')

if p_value < alpha:
    print('Нулевую гипотезу отвергаем: средние пользовательские оценки отличаются.')
else:
    print('Нет оснований отвергнуть нулевую гипотезу: статистически значимых отличий нет.')
```

Вывод

Нулевая гипотеза отвергается и получается,что pc и xone разнятся.

Выдвинем гипотезы:

        a. Средние пользовательские рейтинги жанров Action и Sports равны.
        b. Средние пользовательские рейтинги жанров Action и Sports не равны.
        

```python
action = actual_games.query('genre == "Action" and user_score > -1')
print('Описательная статистика пользовательских рейтингов по Action\n\n', action['user_score'].describe())
sports = actual_games.query("genre == 'Sports' and user_score > -1")
print('Описательная статистика пользовательских рейтингов по Sports\n\n', sports['user_score'].describe())
```

```python

alpha = 0.05

results = st.ttest_ind(action['user_score'], sports['user_score'],equal_var=False,alternative='two-sided')
print(results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу')
    
```

Нулевая гипотеза отвергнута и исходя от противного, есть основания утверждать, что рейтинги жанров Action и Sports не равны.
